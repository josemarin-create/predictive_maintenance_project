# Conclusiones del proyecto

Este notebook resume los principales resultados del proyecto de mantenimiento predictivo.

El objetivo ha sido construir un modelo capaz de predecir si una maquina industrial presenta riesgo de fallo usando variables operativas como temperatura, velocidad de rotacion, torque y desgaste de herramienta.


## Problema planteado

Las empresas manufactureras dependen de maquinaria industrial para mantener su produccion activa. Cuando una maquina falla de forma inesperada, puede provocar paradas de produccion, retrasos, costes de reparacion y perdida de eficiencia.

Este proyecto plantea una primera solucion de mantenimiento predictivo. La idea es usar datos operativos de las maquinas para anticipar posibles fallos y ayudar a priorizar tareas de mantenimiento.


## Dataset utilizado

Se ha utilizado el dataset `AI4I 2020 Predictive Maintenance Dataset`.

El dataset contiene 10000 registros y 14 columnas. La variable objetivo es `Machine failure`, que indica si una maquina ha fallado o no.

Para el modelo principal se han usado variables operativas disponibles antes de conocer el fallo:

- `Type`
- `Air temperature [K]`
- `Process temperature [K]`
- `Rotational speed [rpm]`
- `Torque [Nm]`
- `Tool wear [min]`

No se han utilizado las columnas `TWF`, `HDF`, `PWF`, `OSF` y `RNF` como variables predictoras porque representan tipos concretos de fallo y podrian introducir fuga de informacion.



## Modelos probados

Se compararon varios modelos supervisados de clasificacion:

- Regresion Logistica
- Arbol de Decision
- Random Forest
- Gradient Boosting
- KNN

La comparacion no se baso solo en accuracy, ya que la variable objetivo esta muy desbalanceada. Por ese motivo se dio mas importancia a metricas como precision, recall, F1-score, falsos positivos y falsos negativos para la clase fallo.

Tambien se probo un modelo no supervisado, `Isolation Forest`, para detectar comportamientos anomalos. Este modelo no uso la variable objetivo durante el entrenamiento y se incluyo como analisis complementario.



## Modelo seleccionado

El modelo seleccionado ha sido Gradient Boosting con un umbral de decision ajustado a 0.3.

Con este ajuste, el modelo obtiene para la clase fallo:

- Precision: 0.80
- Recall: 0.82
- F1-score: 0.81

Esto significa que el modelo consigue un equilibrio razonable entre detectar fallos reales y evitar demasiadas falsas alarmas.


## Interpretacion de negocio

El modelo podria ayudar a una empresa industrial a identificar maquinas con mayor riesgo de fallo.

Esto permitiria priorizar revisiones de mantenimiento, reducir paradas inesperadas y usar mejor los recursos tecnicos.

El resultado no debe interpretarse como una solucion lista para produccion, sino como una prueba de concepto. Para aplicarlo en una empresa real seria necesario validarlo con datos historicos propios de sus maquinas.


## Limitaciones

El dataset utilizado es sintetico, por lo que no refleja completamente la complejidad de una fabrica real.

Tambien existe un fuerte desbalance entre maquinas con fallo y sin fallo, algo habitual en mantenimiento predictivo.

Ademas, el modelo se ha entrenado con un numero limitado de variables. En un caso real seria recomendable incorporar mas informacion, como historico de mantenimiento, turnos de trabajo, condiciones ambientales, proveedor de piezas o registros temporales de sensores.

Isolation Forest no obtuvo buenos resultados para detectar fallos directamente, ya que muchos fallos no aparecian como anomalias claras. Por este motivo, no se selecciono como modelo principal.
